```python
!pip install -q statsbombpy mplsoccer
import pandas as pd
from statsbombpy import sb

comps=sb.competitions()
es_pl =comps["competition_name"] == "Premier League"
es_temp = comps["season_name"] == "2015/2016"
pl = comps[es_pl & es_temp]
print(pl)

comp_id = int(pl["competition_id"].iloc[0])
season_id = int(pl["season_id"].iloc[0])
matches = sb.matches(competition_id=comp_id,season_id=season_id)

print("Partidos:", len(matches))
print(sorted(matches["home_team"].unique()))
```

```python
equipos =["Aston Villa","Manchester City","Chelsea","Arsenal","Liverpool"]

local = matches["home_team"].isin(equipos)
visitante = matches["away_team"].isin(equipos)
mis_partidos = matches[local | visitante]

print("Partidos de mis equipos:", len(mis_partidos))
for e in equipos:
  n = ((matches["home_team"] == e)| (matches["away_team"] == e)).sum()
  print(e,"->",n)
mis_partidos[["match_id","match_date","home_team","away_team","home_score","away_score"]].head()
```
